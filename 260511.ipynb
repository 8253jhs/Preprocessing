```python
data = {
    '나이':[22, 35, 14, 55, 42]}
```

```python
df = pd.DataFrame(data)
```

```python
def categorize(data):
    if 0 < data <= 10:
        return "10대 이하"
    elif 10 < data <= 20:
        return "10대"
    elif 20 < data <= 30:
        return "20대"
    else:
        return "30대 이상"
```

```python
df["검사나이"] = df["나이"].apply(categorize)
```

```python
df['프로그램2'] = df['프로그램2'].fillna(df['프로그램2'].mean())
```

```python
df['프로그램3'] = df['프로그램3'].fillna(df['프로그램3'].mean())
```

```python
df['프로그램4'] = df['프로그램4'].fillna(df['프로그램4'].mean())
```

```python
df['평균'] = (df['프로그램1'] + df['프로그램2'] + df['프로그램3'] + df['프로그램4']) / 4
```

```python
def categorize(data):
    if 0 < data <= 55:
        return "하"
    elif 55 < data <= 70:
        return "중"
    else:
        return "상"
```

```python
df["평가분류"] = df["평균"].apply(categorize)
```

```python
filter_df = df[df['평가분류'] == '상']
filter_df
```

```python
filter_df_contains = df[df['평가분류'].str.contains('상', na=False)]
filter_df_contains
```
