```python
sample = sample.drop(['검사코드', '검사결과값(표시)', '성별코드NM', '진료과코드', '진료과코드NM'], axis=1)
```

```python
sample
```

```python
sample['입실일자'] = pd.to_datetime(sample['입실일자'], format='%Y%m%d', errors='coerce')
```

```python
sample['입실시각'] = pd.to_datetime(sample['입실시각'], format='%H%M', errors='coerce').dt.time
```

```python
sample['접수일자'] = pd.to_datetime(sample['접수일자'], format='%Y%m%d', errors='coerce')
```

```python
sample['퇴실일자'] = pd.to_datetime(sample['퇴실일자'], format='%Y%m%d', errors='coerce')
```

```python
sample['퇴실시각'] = pd.to_datetime(sample['퇴실시각'], format='%H%M', errors='coerce').dt.time
```

```python
sample
```

```python
sample['환자ID']=sample['연구번호'].astype("str")
```

```python
sample['환자ID'] = sample['환자ID'].str[0:3] + sample['환자ID'].str[5:]
```

```python
sample
```

```python
sample['검사나이'] = sample['검사나이'].astype("str")
```

```python
sample['검사나이'] = sample['검사나이'].str.rstrip(".0")
```

```python
sample['검사나이'] = sample['검사나이']+"세"
```

```python
sample['입원기간'] = sample['퇴실일자'] - sample['입실일자']
```
